```python
import pandas as pd

# Path to data frame with WSIs
df_exp2 = r"D:\DATA\abmil_exp2.csv"
df_exp2 = pd.read_csv(df_exp2)
df_exp3 = r"D:\DATA\abmil_exp3.csv"
df_exp3 = pd.read_csv(df_exp3)
```

```python
from helper_functions import strings2lists

list_str_cols = ['snomed_code', 'M', 'T', 'snomed_text', 'T_text', 'M_text', 'undersoeger_anonymous', 'T_category', 'M_category']

for col in list_str_cols: 
    df_exp2[col] = df_exp2[col].apply(strings2lists)
    df_exp3[col] = df_exp3[col].apply(strings2lists)    
```

```python
from helper_functions import lists2tuples

df_exp2 = lists2tuples(df_exp2)
df_exp3 = lists2tuples(df_exp3)
```

```python
# Merge dfs 
df_merged = pd.concat([df_exp2, df_exp3], ignore_index=True)

# Remove duplicates
df_merged = df_merged.drop_duplicates()

all_filenames = df_merged["filename"].tolist()
print("Number of wsi filenames: ", len(all_filenames))
```

```python
# Save to csv
output_file = "D:\DATA\priority_slides.csv"

df_merged.to_csv(output_file, index=False)
print(f"Saved DataFrame to {output_file}")
```
