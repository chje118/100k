```python
import pandas as pd

# Path to data frame with WSIs
df_path = "D:\DATA\with_snomed_category.csv"
df_all = pd.read_csv(df_path)
```

```python
all_filenames = df_all["filename"].tolist()
unique_wsi = set(all_filenames)

print("Number of unique wsi filenames: ",  len(unique_wsi))
print("Number of wsi filenames: ", len(all_filenames))
```

```python
from wsi_metadata import WSIMetadata
cache_file = "wsi_metadata_subset_cache.csv"
wsi_meta = WSIMetadata(all_filenames, cache_path=cache_file)
df_meta = wsi_meta.metadata
```

```python
print("Vendor value counts:")
print(df_meta['vendor'].value_counts())

print("\nComment value counts:")
print(df_meta['comment'].value_counts())
```
