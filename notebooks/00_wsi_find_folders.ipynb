```python
import pandas as pd
from wsi_stats import WSIStatsCache

# Path to WSIs (root dir)
ROOT_DIR = "//regsj.intern/appl/Deep_Visual_Proteomics"

# Path to cache file
CACHE_FILE = "D:\DATA\wsi_cache.pkl"

# Load/reload WSI stats cache
wsi_cache = WSIStatsCache(ROOT_DIR, CACHE_FILE)
df_wsi = wsi_cache.main(reload=False)
```

```python
# Check for WSI with no associated data
missing_data = df_wsi[
    (df_wsi["file_size"].isna() | (df_wsi["file_size"] == 0)) |
    (df_wsi["data_folder_size"].isna() | (df_wsi["data_folder_size"] == 0))
]

# Exact prefix with proper escaping
prefix = r"//regsj.intern/appl/Deep_Visual_Proteomics\\"

# Create a new column 'short_name' by stripping the prefix from the index
missing_data["short_name"] = (
    missing_data.index.to_series().str.replace(prefix, '', regex=True)
)

# Print short names as a plain list
print("WSIs with missing data:", len(missing_data))
for name in missing_data["short_name"]:
    print(name)
```

```python
missing_data = missing_data.reset_index()  # filename becomes a column
print(missing_data.head())
```

```python
from wsi_stats import FindWSIData
import os

print("Root directory:", ROOT_DIR)
print("Exists:", os.path.exists(ROOT_DIR))

finder = FindWSIData(ROOT_DIR, df = missing_data, col = 'filename')
df_result = finder.df_results
```

```python
for idx, row in df_result.iterrows():
    filename = row.get("filename")
    matching_folders = row.get("matching_folders")
    print(filename, ": \n", matching_folders)
    print(" ")
```

```python
# Save to csv
output_file = r"D:\DATA\full_dataset\wsi_without_data.csv"
df_result.to_csv(output_file, index=False)

print(f"Saved DataFrame to {output_file}")
```
