```python
import pandas as pd

# Path to main directory
path = r"D:\NOTEBOOKS\Christine\all_slides\combined_feature_summary.csv"
df = pd.read_csv(path)
```

```python
complete = "feature extraction complete"

df_complete = df[df["status"].apply(lambda x: (x == complete))].copy()
```

```python
# Convert "elapsed time" to a timedelta
df_complete["elapsed_time"] = pd.to_timedelta(df_complete["elapsed_time"])

# Compute status count and average elapsed time for each model 
status_counts = df_complete.groupby("model")["status"].value_counts().unstack(fill_value=0)
avg_elapsed = df_complete.groupby("model")["elapsed_time"].mean()

# Convert average timedelta to readable string
avg_elapsed_str = avg_elapsed.apply(lambda x: str(x))

print("Status Counts by Model:")
print(status_counts)

print("\nAverage Elapsed Time by Model (HH:MM:SS):")
print(avg_elapsed_str)
```

```python
import datetime as dt

# Column with elapsed time in minutes
def to_minutes(x):
    try:
        return pd.to_timedelta(x).total_seconds() / 60
    except Exception:
        return None
        
df_complete["elapsed_minutes"] = df_complete["elapsed_time"].apply(to_minutes)
```

```python
print(df_complete.head())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 6))
sns.boxplot(
    data=df_complete,
    x="model",
    y="elapsed_minutes",
    showfliers=True
)
plt.title("Elapsed Time Distribution by Model")
plt.xlabel("Model")
plt.ylabel("Elapsed Time (minutes)")
plt.tight_layout()
plt.show()
```
