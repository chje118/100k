```python
import pandas as pd

# Path to data frame containing all SNOMED counts
df_path = "D:\DATA\snomed_counts.csv"
df_codes = pd.read_csv(df_path)
```

```python
prefix_colors = {
    'M': '#457b9d',
    'T': '#bde0fe',
    'F': '#b7e4c7',
    'P': '#ffd6a5',
    'Æ': '#ffb4a2',
    'S': '#cdb4db',
    'Function': '#b7e4c7',
    'Disease': '#cdb4db',
}
```

```python
print(df_codes.head())
```

```python
import plotly.graph_objects as go
import plotly.express as px
```

```python
# Figure 1
# SNOMED Distribution Sunburst
fig = px.sunburst(
    df_codes,
    path=['prefix', 'category'],
    values='count',
    color='prefix',
    color_discrete_map=prefix_colors,
    title='Distribution of SNOMED Codes', 
    width=600, 
    height=600
)
fig.update_traces(
    rotation=-90,
    textinfo='label+percent entry',
    insidetextorientation='horizontal',
)
fig.show()
```

```python
# Figure 1
# Zoomed Sunburst for SNOMED S and F codes (not visibile in full sunburst) 
zoomed_df = df_codes[df_codes['prefix'].isin(['S', 'F'])].copy()

total_other = zoomed_df['count'].sum()
total_all = df_codes['count'].sum()
percent_other = 100 * total_other / total_all

fig_other_sunburst = px.sunburst(
    zoomed_df,
    path=['category'],
    values='count',
    color='category',  # color by Category for correct outer colors
    color_discrete_map=prefix_colors,
    title='Distribution of other SNOMED Codes (S, F)',
    width=600, 
    height=600
)
fig_other_sunburst.update_traces(
    textinfo='label+percent entry',
    insidetextorientation='horizontal'
)
fig_other_sunburst.add_annotation(
    x=0.5, y=0.5,
    text=f"{percent_other:.1f}%",
    showarrow=False,
    font=dict(size=18, color="black"),
    xref="paper", yref="paper"
)
fig_other_sunburst.show()
```

```python
# Figure 1
# Top Morphology Codes
df_m = df_codes[df_codes['prefix'] == 'M'].copy().sort_values('count', ascending=False).head(10)
max_count = df_m['count'].max()

fig_m = go.Figure()
fig_m.add_trace(go.Bar(
    y=df_m['category'][::-1], 
    x=df_m['count'][::-1],
    orientation='h',
    marker_color=prefix_colors['M'],
    name='Morphology',
    text=df_m['count'][::-1],
    textposition='outside'
))
fig_m.update_layout(
    template="simple_white",
    title="Top 10 Morphology Categories",
    xaxis_title="Count",
    yaxis_title="Category",
    showlegend=False,
    width=900, 
    height=400
)
fig_m.update_xaxes(range=[0, max_count * 1.15])
fig_m.show()
```

```python
# Figure 1 
# Top Topography Codes
df_t = df_codes[df_codes['prefix'] == 'T'].copy().sort_values('count', ascending=False).head(10)
max_count = df_t['count'].max()

fig_t = go.Figure()
fig_t.add_trace(go.Bar(
    y=df_t['category'][::-1],
    x=df_t['count'][::-1],
    orientation='h',
    marker_color=prefix_colors['T'],
    name='Topography',
    text=df_t['count'][::-1],
    textposition='outside'
))
fig_t.update_layout(
    template="simple_white",
    title="Top 10 Topography Categories",
    xaxis_title="Count",
    yaxis_title="Category",
    showlegend=False,
    width=900, 
    height=400
)
fig_t.update_xaxes(range=[0, max_count * 1.15])
fig_t.show()
```
